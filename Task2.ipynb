```python
addr = "1.1.1.1"
res = ''
for i in addr:
    if i =='.':
        res+= "[.]"
    else:
        res+= i

print(res)
```

```python
nums = [1,1,1,1]

k=0
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i]==nums[j]:
            k+=1

print(k)
```

```python
nums = [8,1,2,2,3]

res=[]
for i in range(len(nums)):
    k=0
    for j in range(len(nums)):
        if nums[i]>nums[j]:
            k+=1
    res.append(k)
print(res)
```

```python
n = 234
sum_ = 0
mul = 1

while n > 0:
    digit = n % 10
    sum_ += digit
    mul *= digit
    n //= 10
print(mul-sum_)
```

```python
class Solution(object):
    def xorOperation(self, n, start):
        result = 0
        for i in range(n):
            result ^= start + 2*i
        return result
```
